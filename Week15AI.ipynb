# Personal Task Manager

This notebook contains a Python script for a simple 'Personal Task Manager'. It allows you to add tasks with a name and a priority level (High, Medium, Low), view all added tasks, and exit the program.

```python
class Task:
    PRIORITY_MAP = {
        1: 'High',
        2: 'Medium',
        3: 'Low'
    }

    def __init__(self, name, priority):
        if priority not in self.PRIORITY_MAP:
            raise ValueError("Priority must be 1 (High), 2 (Medium), or 3 (Low).")
        self.name = name
        self.priority = priority

    def __str__(self):
        return f"Task: {self.name}, Priority: {self.PRIORITY_MAP[self.priority]}"

    def __repr__(self):
        return f"Task(name='{self.name}', priority={self.priority})"
```

Next, we'll create a `TaskManager` class to handle the operations related to adding and listing tasks.

```python
class TaskManager:
    def __init__(self):
        self.tasks = []
        self.completed_tasks_file = 'completed_tasks.txt'
        # Ensure the completed tasks file exists
        open(self.completed_tasks_file, 'a').close()

    def add_task(self, name, priority):
        try:
            task = Task(name, priority)
            self.tasks.append(task)
            print(f"Task '{name}' added successfully with {task.PRIORITY_MAP[priority]} priority.")
        except ValueError as e:
            print(f"Error adding task: {e}")

    def list_tasks(self):
        if not self.tasks:
            print("No tasks available.")
            return

        # Sort tasks by priority (1=High, 2=Medium, 3=Low)
        sorted_tasks = sorted(self.tasks, key=lambda task: task.priority)

        print("\n--- Your Tasks ---")
        for i, task in enumerate(sorted_tasks, 1):
            print(f"{i}. {task}")
        print("------------------")

    def complete_task(self, task_index):
        if not self.tasks:
            print("No tasks to complete.")
            return

        try:
            # Adjust index for 0-based list, after sorting
            sorted_tasks = sorted(self.tasks, key=lambda task: task.priority)
            task_to_complete = sorted_tasks[task_index - 1]

            # Remove the task from the original (unsorted) list
            self.tasks.remove(task_to_complete)

            # Log to file
            with open(self.completed_tasks_file, 'a') as f:
                f.write(f"{task_to_complete.name} (Priority: {task_to_complete.PRIORITY_MAP[task_to_complete.priority]})\n")
            print(f"Task '{task_to_complete.name}' marked as complete and removed.")

        except IndexError:
            print("Invalid task number.")
        except Exception as e:
            print(f"An error occurred while completing the task: {e}")
```

Finally, here is the main program loop that interacts with the `TaskManager` class, allowing you to add, view, and exit the application.

```python
def run_task_manager():
    manager = TaskManager()

    while True:
        print("\n--- Task Manager Menu ---")
        print("1. Add a new task")
        print("2. View all tasks")
        print("3. Complete a task")
        print("4. Exit")
        choice = input("Enter your choice: ")

        if choice == '1':
            task_name = input("Enter task name: ")
            while True:
                try:
                    priority_input = input("Enter priority (1 for High, 2 for Medium, 3 for Low): ")
                    priority = int(priority_input)
                    if priority in [1, 2, 3]:
                        manager.add_task(task_name, priority)
                        break
                    else:
                        print("Invalid priority. Please enter 1, 2, or 3.")
                except ValueError:
                    print("Invalid input. Please enter a number for priority.")
        elif choice == '2':
            manager.list_tasks()
        elif choice == '3':
            manager.list_tasks() # Show current tasks to help user choose
            if manager.tasks: # Only ask for input if there are tasks to complete
                while True:
                    try:
                        task_num_input = input("Enter the number of the task to complete: ")
                        task_num = int(task_num_input)
                        manager.complete_task(task_num)
                        break
                    except ValueError:
                        print("Invalid input. Please enter a number.")
        elif choice == '4':
            print("Exiting Task Manager. Goodbye!")
            break
        else:
            print("Invalid choice. Please try again.")

# Run the task manager application
run_task_manager()
```
