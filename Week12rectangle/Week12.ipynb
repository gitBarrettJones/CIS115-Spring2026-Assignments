```python
# NAME: rectangle.py
# IMPORT STATEMENTS

# FUNCTION DEFINITIONS
## define calculate_rectangle parameters length & width ##
def calculate_rectangle(length, width):
# find area using formula length * width #
    area = length * width
# find perimeter using formula 2(length + width) #
    perimeter = 2 * (length + width)
# return area and perimeter #
    return area, perimeter

# PROGRAM LOGIC
## User input ##
# ask user for length and width using float()#

## call the calculate_rectangle function ##
# unpack the tuple as two variable my_area, my_perimeter #

## print the results ##
calculate_rectangle(5, 10)
```
