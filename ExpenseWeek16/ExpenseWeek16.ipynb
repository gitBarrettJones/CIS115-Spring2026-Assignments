```python
# IMPORT STATEMENTS
# (None)


# FUNCTION DEFINITIONS
## create a list to store records ##

expenses = []


## def the expense to be added ##
def add_expense():
# print a message for user to add expense #
  print("--Add Expense--")
# ask the user to input a Category #
  category = input("Enter a Category: ")
# ask the user to input a Vendor #
  vendor = input("Enter a Vendor: ")
# ask the user to input a amount (use float) #
  while True:
    try:
      amount = float(input("Enter an Amount: "))
      break
    except ValueError:
      print("Invalid input. Please enter a valid number.")
# ensure user is only able to input a number for amount (loop) #
# ask the user to input a date mm-dd-yyyy #
  date = input("Enter a Date (mm-dd-yyyy): ")
# print the user a message after adding the expense #
## create dictionary for storage of expenses ##
  expense = {
    "Category": category,
    "Vendor": vendor,
    "Amount": amount,
    "Date": date
  }
  expenses.append(expense)
  print("Expense added successfully!")
## def total expenses ##
def total_all_expenses():
    total = sum(exp["Amount"] for exp in expenses)
    print(f"Total of all expenses: ${total:.2f}")

## def total category ##
def total_category():
# ask the user to input a category to see #
    category = input("Enter a category: ")
# for any matching category display the total amount within #
    total = sum(exp["Amount"] for exp in expenses if exp["Category"] == category)
    print(f"Total of {category}: ${total:.2f}")
## def total vendor ##
def total_vendor():
# ask the user to input a vendor #
    vendor = input("Enter a vendor: ")
# for any matching vendor display the total amount within #
    total = sum(exp["Amount"] for exp in expenses if exp["Vendor"] == vendor)
    print(f"Total of {vendor}: ${total:.2f}")
## def the menu to select options ##
def menu():
# display menu option #
# add expense, view total all, view total by category, view total by vendor, exit #
  print("---Expense Tracker---")
  print("1. Add Expense")
  print("2. View Total of All Expenses")
  print("3. View Total by Category")
  print("4. View Total by Vendor")
  print("5. Exit")



# PROGRAM LOGIC

## loop program ##
## call the menu ##
while True:
  menu()
  choice = input("Enter your choice: ")
# ask the user to input an option #
# for the selected option picked call the corresponding function #
  if choice == "1":
    add_expense()
  elif choice == "2":
    total_all_expenses()
  elif choice == "3":
    total_category()
  elif choice == "4":
    total_vendor()
  elif choice == "5":
    print("Exiting the program.")
    break
  else:
    print("Invalid choice. Please try again.")
```
