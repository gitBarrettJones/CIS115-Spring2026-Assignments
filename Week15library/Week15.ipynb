```python
# NAME: library.py
# IMPORT STATEMENTS

# CLASS DEFINITIONS
## class: Book ##
class Book:
# def _init_(self, title, author) #
# use self. for title, author, and is_available = true #
  def __init__(self, title, author):
    self.title = title
    self.author = author
    self.is_available = True

## def _str_() ##
# if the book is available = true #
  def __str__(self):
# produce a string saying it is available #
    if self.is_available:
      status = "Available"
# else produce a string saying 'checked out'#
    else:
      status = "Checked Out"
    return f"{self.title} by {self.author} - {status}"

# def checkout() to is_available = false #
  def checkout(self):
    self.is_available = False
# def return_book that set is_available to = true #
  def return_book(self):
    self.is_available = True
# PROGRAM LOGIC

## Create two instances of the Book class ##
# create Book 1 with a title and author #
my_book = Book("The Great Gatsby", "F. Scott Fitzgerald")
# create Book 2 with a title and author
my_book2 = Book("Mistborn: The Final Empire", "Brandon Sanderson")

## print the status of these Books ##
print("--Library Status--")
print(my_book)
print(my_book2)
## check out a Book using method ##
print("\nChecking out 'The Great Gatsby' by F. Scott Fitzgerald\n")
my_book.checkout()

## print both Books showing the updated status ##
print("New Status:")
print(my_book)
print(my_book2)
```
