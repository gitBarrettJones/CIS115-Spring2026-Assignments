```python
# IMPORT STATEMENTS

# FUNCTION DEFINITIONS

## define the add_students function ##

def add_students(gradebook):
# use while loop #
  while True:
# ask for students name input #
    name = input("Enter student name (or 'done' to quit): ")
# if input is "done" end the loop #
    if name == "done":
      break
    try:
# ask for the students grade input between 0-100 #
# add the name as key and grade is the value #
# else provide a error message #
      grade = float(input(f"Enter grade (0-100) for student {name}: "))
      if 0 <= grade <= 100:
        gradebook[name] = grade
# else #
      else:
        print("Invalid grade. Grade must be between 0 and 100.")
    except ValueError:
      print("Invalid input. Please enter a numerical grade.")


## define the fruitful calculate_average function ##
def calculate_average(gradebook):
  if not gradebook:
    return 0
# calculate all dictionary using .values() #
# return the avarage #
  total_grades = sum(gradebook.values())
  return total_grades / len(gradebook)


# PROGRAM LOGIC

## Create open dictionary ##
# call the add_students to populate dictionary #
def sga():
  gradebook = {}
  add_students(gradebook)

## Part 2: print to verify
  print("\nGradebook:")
  for name, grade in gradebook.items():
    print(f"{name}: {grade}")

  ## Provide final report ##
  # list all student names and grades #
  print("\n--- Class Report ---")
  if not gradebook:
    print("No students in the gradebook.")
  else:
    for name, grade in gradebook.items():
      print( name, grade )

  ## call the calculate_average ##
  # calculate class average and print #
  average_grade = calculate_average(gradebook)
  print(f"\nClass Average: {average_grade:.2f}")

sga()
```
