```python
# NAME: journal.py
# IMPORT STATEMENTS

# FUNCTION DEFINITIONS

## define the write_entry function ##
def write_entry(filename, entry):
# "filename" and "entry" as the parameteres #
# start each file entry on a new line #
# close the file with .close()#
    file = open(filename, 'a')
    file.write(entry + '\n')
    file.close()


## define read_journal ##
def read_journal(filename):
# open the file in read mode with 'r' #
  try:
    file = open(filename, 'r')
    contents = file.read()
    file.close()

# print the contents within to the user #
    if contents.strip() == '':
            print("Your journal is empty.")
    else:
            print("\n--- Your Journal ---")
            print(contents.strip())

  except FileNotFoundError:
        print("No journal found. Start writing one!")


# PROGRAM LOGIC

def main():
    filename = "journal.txt"

    while True:
        action = input("(W)rite, (R)ead, or (Q)uit? ").strip().lower()

        if action == 'q':
            print("Goodbye!")
            break
        elif action == 'w':
            entry = input("Enter your entry: ")
            write_entry(filename, entry)
            print("Entry saved.\n")
        elif action == 'r':
            read_journal(filename)
        else:
            print("Invalid choice. Please type W, R, or Q.\n")

# Run the program
if __name__ == "__main__":
    main()
```
