```python
# NAME: journal.py
# IMPORT STATEMENTS

# FUNCTION DEFINITIONS

## define the write_entry function ##
def write_entry(filename, entry):
# "filename" and "entry" as the parameteres #
# start each file entry on a new line #
# close the file with .close()#
    with open(filename, 'a') as file:
        file.write(entry + '\n')
        file.close()


## define read_journal ##
# open the file in read mode with 'r' #
# print the contents within to the user #
# closing the file with .close #

# PROGRAM LOGIC

## Write a new entry with " w " or " r" ##
# place within a loop with while until the inputs ' quit '#
# when " w " is selected call for ' write_entry ' #
# when reading the journal call ' r ' read_journal #

## testing midpoint ##
def main():
  filename = "journal.txt"

while True:
  choice = input("\nDo you want to (W)rite a new entry or type 'quit' to exit? ").strip().lower()

  if choice == 'w':
            entry = input("Enter your thoughts: ")
            write_entry(filename, entry)
            print("Entry saved!")

  elif choice == 'quit':
            print("Goodbye!")
            break

  else:
            print("Invalid choice. Please type 'W' or 'quit'.")

if __name__ == "__main__":
    main()
```
