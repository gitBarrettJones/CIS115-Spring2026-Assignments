```python
# NAME: roster.py
# IMPORT STATEMENTS

# FUNCTION DEFINITIONS

## define the add_players function ##
def add_players(player_list):
  user_input = ''
# inside use 'while function' #
# ask user to enter name #
# accept user input() #
# have the loop end when input is ' quit '#
# .append to add element to the end of list #
  while user_input != 'quit':
    user_input = input('Enter the name of a player (or type quit to stop): ')
    if user_input != 'quit':
      player_list.append(user_input)

## define the is_on_roster function ##
# fruitful needs to use 'return' #
#'if' the player name is selected, true #
# otherwise false #
def is_on_roster(player_list, search_name):
  if search_name in player_list:
    return True
  else:
    return False

# PROGRAM LOGIC

## empty list : roster[] ##
roster = []
# fill the roster : add_players() #
add_players(roster)
## search for specific name ##
# user input() ask for a name to enter #
search_name = input('Enter a name to search for: ')
# gather results and call is_on_roster #
# if / else to print the results #
results = is_on_roster(roster, search_name)

if results:
  print(search_name, 'is on the roster.')
else:
  print(search_name, 'is NOT on the roster.')
```
